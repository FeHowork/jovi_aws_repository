```sql
CREATE OR REPLACE TABLE jovi_lakehouse.silver.vendas_enriquecidas AS
SELECT
    v.idVenda,
    v.data_venda,
    v.horario,

    v.idPromotor,
    p.nome_colaborador AS promotor,
    p.status AS status_promotor,
    p.uf AS uf_promotor,

    v.idSupervisor,
    s.nome_supervisor,

    v.idLoja,
    l.nome_loja,
    l.rede_loja,
    l.cidade AS cidade_loja,
    l.estado AS estado_loja,

    v.idProduto,
    pr.nome_produto,
    pr.tipo,
    pr.memoria,
    pr.rede,

    v.imei,
    v.vchat

FROM jovi_lakehouse.silver.vendas v

LEFT JOIN jovi_lakehouse.bronze.promotores p
    ON p.id = v.idPromotor

LEFT JOIN jovi_lakehouse.bronze.supervisores s
    ON s.idSupervisor = v.idSupervisor

LEFT JOIN jovi_lakehouse.bronze.lojas l
    ON l.idLoja = v.idLoja

LEFT JOIN jovi_lakehouse.bronze.produtos pr
    ON pr.idProduto = v.idProduto;
```
