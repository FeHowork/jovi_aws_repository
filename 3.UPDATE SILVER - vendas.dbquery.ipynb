```sql
CREATE OR REPLACE TABLE jovi_lakehouse.silver.vendas AS
SELECT
    idVenda,
    CAST(data_venda AS DATE) AS data_venda,
    horario,
    idSupervisor,
    idPromotor,
    idLoja,
    idProduto,
    imei,
    UPPER(TRIM(cidade)) AS cidade,
    UPPER(TRIM(estado)) AS estado,
    vchat
FROM jovi_lakehouse.bronze.vendas
WHERE idVenda IS NOT NULL;
```
